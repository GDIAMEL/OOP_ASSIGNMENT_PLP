```python
# CLASS OF MY CHOICE  (Device)
class Device:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def power_on(self):
        print(f"{self.model} is now powered on!")

    def power_off(self):
        print(f"{self.model} is now powered off.")

# DERIVED CLASS (Smartphone)
class Smartphone(Device):
    def __init__(self, brand, model, battery_level):
        super().__init__(brand, model)  # Initialize base class attributes
        self.battery_level = battery_level  # Specific attribute for Smartphone

    def check_battery(self):
        print(f"Battery level: {self.battery_level}%")

    def charge_phone(self):
        self.battery_level = 100
        print(f"{self.model} is fully charged!")

    def make_call(self, number):
        print(f"Calling {number} from {self.model}...")

# Creating an object of Smartphone
my_phone = Smartphone("Apple", "iPhone 16 Pro Max", 85)
my_phone.power_on()
my_phone.check_battery()
my_phone.make_call("+254713442011")
my_phone.charge_phone()
my_phone.check_battery()
```

```python
# Base class (Vehicle)
class Vehicle:
    def move(self):
        pass  # We'll leave this method to be overridden by subclasses

# Derived class (Car)
class Car(Vehicle):
    def move(self):
        print("Driving")

# Derived class (Plane)
class Plane(Vehicle):
    def move(self):
        print("Flying")

# Derived class (Bicycle)
class Bicycle(Vehicle):
    def move(self):
        print("Pedaling")

# Create instances of each class
vehicle1 = Car()
vehicle2 = Plane()
vehicle3 = Bicycle()

# Call the move() method on each object
vehicle1.move()  # Output: Driving
vehicle2.move()  # Output: Flying
vehicle3.move()  # Output: Pedaling
```
